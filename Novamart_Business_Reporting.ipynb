```python
## AUTOMATION ##
# Put our KPI calculations together into one reusable block.

import pandas as pd

df = pd.read_csv('/content/novamart_ecommerce_sales.csv')

df['order_status'] = df['order_status'].str.strip().str.title()

df['order_date'] = pd.to_datetime(df['order_date'])

df['revenue'] = df['quantity'] * df['unit_price']

completed_df = df[df['order_status'] == 'Completed']
returned_df = df[df['order_status'] == 'Returned']

total_revenue = df['revenue'].sum()
completed_revenue = completed_df['revenue'].sum()
returned_value = returned_df['revenue'].sum()

completed_orders = completed_df['order_id'].nunique()
returned_orders = returned_df['order_id'].nunique()

net_sales = completed_revenue - returned_value
return_rate = (returned_orders / completed_orders) * 100

print("NOVAMART BUSINESS REPORT")
print("------------------------")
print(f"Total Revenue: KSh {total_revenue:,.2f}")
print(f"Completed Revenue: KSh {completed_revenue:,.2f}")
print(f"Estimated Net Sales: KSh {net_sales:,.2f}")
print(f"Completed Orders: {completed_orders:,}")
print(f"Returned Orders: {returned_orders:,}")
print(f"Return Rate: {return_rate:.2f}%")
```

**Role:** Act as a senior business analyst preparing a management report for an e-commerce company.

**Format:** Present an executive summary, three key findings, three management recommendations, and areas requiring further investigation.

```python
def generate_report(file_path):

    import pandas as pd

    df = pd.read_csv(file_path)

    # Clean text columns
    text_columns = ['product', 'region', 'sales_channel', 'order_status']

    for col in text_columns:
        df[col] = df[col].fillna('Unknown').str.strip().str.title()

    # Convert date
    df['order_date'] = pd.to_datetime(df['order_date'])

    # Calculate revenue
    df['revenue'] = df['quantity'] * df['unit_price']

    # Separate order statuses
    completed = df[df['order_status'] == 'Completed']
    returned = df[df['order_status'] == 'Returned']

    # Calculate KPIs
    total_revenue = df['revenue'].sum()
    completed_revenue = completed['revenue'].sum()
    returned_value = returned['revenue'].sum()

    completed_orders = completed['order_id'].nunique()
    returned_orders = returned['order_id'].nunique()

    return_rate = (
        returned_orders / (completed_orders + returned_orders) * 100
        if completed_orders + returned_orders > 0 else 0
    )

    # Display report
    print("NOVAMART BUSINESS REPORT")
    print("------------------------")
    print(f"Total Revenue: KSh {total_revenue:,.2f}")
    print(f"Completed Revenue: KSh {completed_revenue:,.2f}")
    print(f"Returned Value: KSh {returned_value:,.2f}")
    print(f"Completed Orders: {completed_orders:,}")
    print(f"Returned Orders: {returned_orders:,}")
    print(f"Return Rate: {return_rate:.2f}%")
```

```python
# Testing the workflow

generate_report('/content/novamart_ecommerce_sales.csv')

print("REPORT GENERATED SUCCESSFULLY!")
```
